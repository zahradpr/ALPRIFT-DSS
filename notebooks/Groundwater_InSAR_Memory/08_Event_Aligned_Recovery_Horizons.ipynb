## 2. Stage 23 — Event-aligned analysis

The groundwater and InSAR histories were aligned relative to recovery-related events rather than interpreted using one universal basin-wide lag.

This approach was motivated by the earlier finding that fixed-lag correlation and distributed-lag models did not reveal a consistent response time across the study area.

## 3. Stages 24–25 — Non-circular fixed evaluation horizons

Post-recovery subsidence was evaluated at fixed future horizons without circular shifting.

The strongest consistent evidence emerged at approximately **18–42 months**, with longer horizons explored up to the extended analysis range.

These horizons are evaluation times after recovery events; they must **not** be interpreted as a uniquely identified physical lag.

## 4. Stage 26 — First sustained recovery event

For wells satisfying the sustained-recovery definition, continued positive subsidence remained detectable after groundwater recovery.

Key historical results:

| Horizon | N | Median additional subsidence | Positive wells | Sign-test FDR q | Wilcoxon FDR q |
|---|---:|---:|---:|---:|---:|
| 18 months | 15 | 18.56 mm | 86.7% | 0.0111 | 0.0013 |
| 24 months | 16 | 23.46 mm | 81.3% | 0.0160 | 0.0013 |
| 36 months | 11 | 28.99 mm | 100% | 0.0029 | 0.0013 |
| 48 months | 7 | 34.29 mm | 100% | 0.0156 | 0.0117 |

The appropriate interpretation is:

> Evidence of continued or residual subsidence becomes detectable by the **18-month evaluation horizon** and remains FDR-supported through **48 months** in the sustained-recovery subset.

This is **not equivalent to saying that the groundwater–subsidence lag is 18 months**.

## 5. Final scientific interpretation

Across the full workflow, no robust universal fixed lag was identified.

Instead, the cumulative, recovery-event and fixed-horizon analyses show that land subsidence can continue after groundwater levels begin to recover.

The observations are compatible with **hydromechanical memory and residual compaction**, but the available observational evidence does not by itself prove one unique deformation mechanism.

This distinction is important for integration with the later dynamic ALPRIFT / BME framework.
