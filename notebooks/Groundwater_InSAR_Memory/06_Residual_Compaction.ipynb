## 2. Stage 17 — Post-minimum-head behaviour

The post-minimum-head analysis evaluates whether subsidence continued after groundwater heads had begun to recover.

For wells with at least 12 months of post-minimum follow-up:

- **23 wells** satisfied the follow-up requirement;
- all 23 showed groundwater recovery;
- **20 of 23** nevertheless accumulated additional positive stitched subsidence.

This pattern is consistent with continued or residual compaction after hydraulic recovery has begun.

## 3. Stage 18 — Strong residual-compaction candidates

The historical strong-candidate screening used:

- minimum follow-up ≥ 12 months;
- groundwater recovery ≥ 0.5 m;
- additional stitched subsidence ≥ 10 mm.

This produced **11 strong candidates** for further validation.

## 4. Stage 19 — Within-stack validation

The strong candidates were then checked using within-stack information to reduce the possibility that apparent post-recovery displacement was created by SBAS stack-boundary stitching.

The validation retained **10 of the 11 candidates**.

## 5. Scientific interpretation

These results do not imply a universal fixed groundwater–subsidence lag.

Instead, they provide evidence that subsidence can continue after groundwater-level recovery has begun. This behaviour is compatible with hydromechanical memory or residual compaction, while the observational analysis alone does not prove a unique underlying mechanism.
